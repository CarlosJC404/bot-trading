```python
import pandas as pd
import json
import numpy as np
import requests
```

```python
url = 'http://127.0.0.1:6000/api/data/get?badge=NZDUSD&frame=H4&from=2023-11-20 00:00:00&to=2023-11-24 00:00:00'  # URL a la que deseas hacer la solicitud
response = requests.get(url)  # Realizar la solicitud GET

# Verificar si la solicitud fue exitosa (código de respuesta 200)
if response.status_code == 200:
    data = response.json()  # Convertir la respuesta JSON a un diccionario de Python
    df = pd.DataFrame(data)
    df['time']=pd.to_datetime(df['time'], unit='s')
    print(df[["time", "open", "close", "low", "high"]])
    
else:
    print("La solicitud no fue exitosa. Código de respuesta:", response.status_code)
```
