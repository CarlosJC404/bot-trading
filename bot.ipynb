```python
import json
import MetaTrader5 as mt5
from datetime import datetime
import pandas as pd
import pytz
```

```python
class Bot:
    def __init__(self):
        self.credentials = json.load(open("./settings/credentials.json", "r"))["MetaTrader5"]
        self.settings = json.load(open("./settings/bot.json", "r"))

    def connect_MetaTrader5(self):
        settings = {
            "path": "C:\\Program Files\\MetaTrader 5\\terminal64.exe",
            "timeout": 60000,
            "portable": False
        }
        if mt5.initialize(path=settings["path"], login=self.credentials["login"], password=self.credentials["password"], server=self.credentials["server"], timeout=settings["timeout"], portable=settings["portable"]):
            print("Conectado!")
        else:
            print(f"Ha habido un problema en la inicialización: {mt5.last_error()}")
    
    def get_data(self):
        badge = "NZDUSD"
        time_frame = mt5.TIMEFRAME_D1

        # Fecha desde donde se recuperaran los datos
        timezone = pytz.timezone("Etc/UTC")
        date = datetime(2023, 12, 31)
        print(date)

        raw_ohlc_data = mt5.copy_rates_from(badge, time_frame, date, 60)
        df = pd.DataFrame(raw_ohlc_data)
        df['time'] = pd.to_datetime(df['time'], unit='s')
        print(df[["time", "open", "close", "low", "high"]])
        
        
        # with open("./data/NZDUSD-M15.json", "w") as file:
        #     json.dump(df.to_dict(orient="records"), file)
```
