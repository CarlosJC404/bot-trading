```python
import datetime

fecha = datetime.datetime(2023, 12, 16)
last = datetime.datetime(2023, 7, 22) 

tiempo_unix = int(fecha.timestamp())
last_unix = int(last.timestamp())

print(tiempo_unix / 7)
print(last_unix / 7)

# print(f"Tiempo Unix para {fecha}: {tiempo_unix - last_unix} segundos desde 1970.")
```
